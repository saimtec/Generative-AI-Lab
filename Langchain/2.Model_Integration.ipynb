# Integration of Ollama, Gemini, and Groq

We are going to learn how to integrate **Ollama, Gemini, and Groq** with LangChain.

---

## Ways to Integrate Models

There are two common approaches for integrating models in LangChain:

### 1. Using `init_chat_model()`

The `init_chat_model()` approach provides a **unified interface** for initializing different chat models.

It allows you to switch between different model providers while keeping a similar interface.

**Example providers:**

- Ollama
- Gemini
- Groq
- OpenAI
- Anthropic



### 2. Using Provider-Specific Classes

The second approach is to use the **provider-specific integration class**, such as `ChatGroq`.

For example:

- `ChatGroq` → Groq models
- `ChatGoogleGenerativeAI` → Gemini models
- `ChatOllama` → Ollama models

This approach gives you direct access to the integration for a specific provider.



```python
# Gemini integration thorugh init method

import os
from langchain.chat_models import init_chat_model
os.environ["GOOGLE_API_KEY"] = os.environ.get("GOOGLE_API_KEY")

model = init_chat_model(
    "google_genai:gemini-3.5-flash-lite",
)

response = model.invoke("Hello, how are you?")
print(response.content)
```

```python
# google gemini specific integration 

from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
response = model.invoke("Hello, how are you?")
print(response.content)
```

```python
# ollama integration 

import os
from dotenv import load_dotenv
from langchain_ollama import ChatOllama

load_dotenv()

model = ChatOllama(
    model = os.getenv("OLLAMA_MODEL"),
    base_url = os.getenv("OLLAMA_BASE_URL"),
)

response = model.invoke("Hello, how are you?")
print(response.content)
```

```python
# Groq integration 

import os 
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.environ.get("GROQ_API_KEY")


model = init_chat_model("groq:qwen/qwen3.8-27b", max_tokens=256)

response = model.invoke("Hello engine")
print(response.content)
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="qwen/qwen3.8-27b")
response = model.invoke("Hello engine", max_tokens=256)
print(response.content)
```

## 1. Streaming

**Streaming** means receiving the model's response **chunk by chunk** as it is generated instead of waiting for the complete response.



```python
for chunk in model.stream("Explain machine learning"):
    print(chunk.text, end="")
```

## Batch Processing

**Batch processing means sending multiple inputs to the model together and processing them as a batch.**

```python
response = model.batch([
    "What is the Capital of France?",
    "What is the Capital of Germany?",
    "What is Deep Learning?"
])

for r in response:
    print(r.text)
```
