# Conversation Memory in LangChain

Conversation memory allows an AI to use previous messages to maintain context and provide relevant responses during a conversation.

---

# 1. Conversation Buffer Memory

Conversation Buffer Memory stores the **entire raw conversation history**, including every human message and AI response, and passes it to the model during future interactions.

**How it works**

Each exchange is appended to the message history.

```text
User Message 1
      ↓
AI Response 1
      ↓
User Message 2
      ↓
AI Response 2
      ↓
Full Conversation History → LLM




**Advantages**

- Simple to understand and implement.
- Preserves the original messages without summarizing them.
- Provides detailed conversational context.
- Helps the model answer questions about specific earlier statements.
- Avoids information loss caused by summarization.

**Disadvantages**

- Conversation history grows with every interaction.
- Consumes more tokens and can increase API costs.
- May eventually exceed the model's context window.
- Longer prompts can increase response latency.
- Repeatedly sending the same history can be inefficient.

````python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_classic.chains.conversation.base import ConversationChain
from langchain_classic.memory import ConversationBufferMemory


load_dotenv()


model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
````

````python
memory = ConversationBufferMemory()

chain = ConversationChain(
    llm=model,
    memory=memory,
    verbose=True,
)

print(chain.predict(input="Hi, my name is Sam."))
````

````python
print(chain.predict(input="How are yoou doing?"))
````

````python
print(chain.predict(input="What do you think about Pakistan?"))
````

# 3. Conversation Summary Buffer Memory

Conversation Summary Buffer Memory combines Conversation Buffer Memory and Conversation Summary Memory.

It keeps recent messages in their original form while summarizing older messages when the conversation becomes too long.

**How it works**
Older Conversation Messages
            ↓
        LLM Summary
            ↓
  Summary of Older Messages
            +
    Recent Raw Messages
            ↓
           LLM


**Advantages**

- Preserves recent messages exactly.
- Retains important information from older conversations through summaries.
- Uses less prompt space than keeping the complete history indefinitely.
- Balances detailed recent context with long-term conversational context.
- Useful for assistants that handle long, ongoing conversations.

**Disadvantages**

- Requires summarization processing.
- Older details may be lost or simplified.
- More complex than basic buffer memory.
- Requires a strategy for deciding when to summarize and how much history to retain.
- Still consumes tokens for both the summary and recent messages.

````python
from langchain_classic.memory import ConversationSummaryMemory

memory = ConversationSummaryMemory(llm=model)

conversation = ConversationChain(
    llm=model,
    memory=memory,
    verbose=True,
)

````

````python
conversation.predict(input="Hi, my name is Sam.")
````

````python
conversation.predict(input="Can you help me in writing a Research Paper?")
````

````python
print(conversation.memory.buffer)
````

# 4. Conversation Buffer Window Memory

Conversation Buffer Window Memory stores only the **most recent \(k\) conversation exchanges** and removes older messages from the active history. This helps limit the amount of context passed to the LLM.

**How it works**

If `k = 2`, the memory keeps the latest two conversation exchanges instead of the entire history.

```text
Conversation History

Exchange 1 → Removed
Exchange 2 → Removed
Exchange 3 → Kept
Exchange 4 → Kept
                 ↓
          Recent History → LLM

````python
from langchain_classic.memory import ConversationBufferWindowMemory

window_memory = ConversationBufferWindowMemory(k=2)   # k is the number of previous interactions to keep in memory

conversation = ConversationChain(
    llm=model,
    memory=window_memory,
    verbose=True,
)
````

````python
conversation.predict(input="Hi, Tell me about python keep it short.")
````

````python
conversation.predict(input="What is the capital of France?")
````

````python
conversation.predict(input="Who is the president of the Pakistan?")
````

````python
conversation.predict(input="Hi, my name is Sam.")
````

# 5. Conversation Summary Buffer Memory

Conversation Summary Buffer Memory combines **Conversation Buffer Memory** and **Conversation Summary Memory**. It keeps recent messages in their original form while summarizing older messages to reduce the amount of conversation history passed to the LLM.

**How it works**

```text
Older Conversation Messages
            ↓
        LLM Summary
            ↓
  Summary of Older Messages
            +
    Recent Raw Messages
            ↓
           LLM



````python
from langchain_classic.memory import ConversationSummaryBufferMemory

memory = ConversationSummaryBufferMemory(llm=model, max_token_limit=500)

conversation = ConversationChain(
    llm=model,
    memory=memory,
    verbose=True,
)
````

````python
conversation.predict(input="Hi, my name is Sam.")
````

````python
conversation.predict(input="Tell me about the Computer vision.")
````

````python
conversation.predict(input="Tell me about the Lahore University.")
````

````python
conversation.predict(input="Hi, How are you?")
````

# Entity Memory

Entity Memory stores important information about specific entities, such as people, companies, places, or projects. It extracts and updates facts about these entities so the LLM can use them in future conversations.

**How it works**

The system identifies entities and stores relevant facts about each one.

Example conversation:

User: My name is Ali, and I work at Microsoft.

User: I am a Python developer.

Stored Memory:

Ali:
- Works at Microsoft.
- Is a Python developer.

Later:

User: Where do I work?

AI: You work at Microsoft.

**Advantages**

- Remembers important facts about specific entities.
- Helps personalize future responses.
- Avoids repeatedly asking for the same information.
- Organizes information by entity.

**Disadvantages**

- May extract or store incorrect facts.
- Requires a strategy for updating outdated information.
- Can consume storage as the number of entities grows.
- May miss relationships between multiple entities.

**Example**

```python
entity_memory = {
    "Ali": {
        "company": "Microsoft",
        "skill": "Python developer"
    }
}

print(entity_memory["Ali"]["company"])
# Microsoft

````python
from langchain_classic.memory import ConversationEntityMemory
from langchain_core.prompts import PromptTemplate

entity_memory = ConversationEntityMemory(llm=model)

entity_prompt = PromptTemplate(
    input_variables=["entities", "history", "input"],
    template=(
        "You are a helpful assistant. Use the entity information and "
        "conversation history when answering.\n\n"
        "Entity information:\n{entities}\n\n"
        "Conversation history:\n{history}\n"
        "Human: {input}\n"
        "AI:"
    ),
)

entity_conversation = ConversationChain(
    llm=model,
    memory=entity_memory,
    prompt=entity_prompt,
    verbose=True,
)

print(entity_conversation.predict(input="My name is Saim, and I work at Microsoft."))
print(entity_conversation.predict(input="I am a Python developer."))
print(entity_conversation.predict(input="Where do I work, and what do I do?"))
````

# 2. Knowledge Graph Memory


Knowledge Graph Memory stores information as **entities and relationships** between them. Instead of storing isolated facts, it represents how different entities are connected.

**How it works**

Information is stored as triples:

$$
(\text{Subject}, \text{Relationship}, \text{Object})
$$

**Example**

User: Ali works at Microsoft and uses Python.

Knowledge Graph:

Ali ── works at ──> Microsoft

Ali ── uses ──> Python

The system can use these relationships to answer questions about Ali, his company, or his skills.

**Advantages**

- Represents relationships between multiple entities.
- Helps answer questions that require connecting facts.
- Organizes information in a structured form.
- Useful for knowledge-based assistants and complex queries.

**Disadvantages**

- More complex to build and maintain than simple memory.
- Incorrect relationships can lead to incorrect answers.
- Requires strategies for updating and removing outdated facts.
- Large knowledge graphs can require specialized storage and querying systems.

````python
from langchain_community.memory.kg import ConversationKGMemory

kg_memory = ConversationKGMemory(llm=model)

kg_conversation = ConversationChain(
    llm=model,
    memory=kg_memory,
    verbose=True,
)

print(kg_conversation.predict(input="Saim works at Microsoft and uses Python."))
print(kg_conversation.predict(input="Where does Saim work, and what does he use?"))
````
