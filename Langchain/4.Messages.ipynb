````python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
````

# Messages in LangChain

**Messages** are the fundamental unit of conversation context in **LangChain**. They represent different parts of a conversation and carry information between the user, model, and tools.

---

# Types of Messages

## 1. System Message

A **System Message** provides instructions to the model about how it should behave.

It is commonly used to define the model's:

- Role
- Behavior
- Rules
- Instructions

### Example

```python
from langchain_core.messages import SystemMessage

message = SystemMessage(
    content="You are a helpful Python teacher."
)

print(message)

## 2. Human Message

**A Human Message represents the user's input or question.**

````python
from langchain_core.messages import HumanMessage

message = HumanMessage(
    content="What is a Python list?"
)

print(message)
````

## 3. AI Message

### An AI Message represents a response generated by the AI model.

**It can contain:**

* Text response
* Tool calls
* Metadata



````python

from langchain_core.messages import AIMessage

message = AIMessage(
    content="A Python list is a collection of items."
)

print(message)
````

## 4. Tool Message

### A Tool Message represents the result returned by a tool after the AI requests that tool.




````python
from langchain_core.messages import ToolMessage

message = ToolMessage(
    content="The weather is 30°C and sunny.",
    tool_call_id="weather_123"
)

print(message)
````

## Passing Messages to the Model

**The message history can be passed to the model using model.invoke().**
**The model uses the previous messages to understand the conversation context.**

````python
from langchain_core.messages import (
    HumanMessage,
    AIMessage
)

messages = [
    HumanMessage(
        content="My name is Saim."
    ),

    AIMessage(
        content="Nice to meet you, Saim!"
    ),

    HumanMessage(
        content="What is my name?"
    )
]

response = model.invoke(messages)

print(response.text)
````
