# Model Context Protocol (MCP)

## 1. What is MCP?

**MCP (Model Context Protocol)** is an open protocol that allows AI applications to connect with external tools, data sources, and services through a standardized interface.

Think of MCP as a common communication protocol between AI applications and external systems.

For example, an AI chatbot might need to:

- Read files from your computer.
- Query a database.
- Search the web.
- Access GitHub repositories.
- Send emails.
- Interact with external APIs.

Without MCP, developers often need to build separate integrations for different applications and services.

MCP provides a standardized way to expose these capabilities.

**Simple example:**

Imagine you build an AI assistant using LangGraph. You want it to access GitHub, read files, and query a database.

Instead of implementing every integration from scratch, you can connect to MCP servers that expose these capabilities.

---

## 2. Why Do We Need MCP?

Suppose you have three AI applications:

- Chatbot A (LangGraph)
- Chatbot B (Claude)
- Chatbot C (Another AI application)

And three external services:

- GitHub
- Database
- File System

Without a standardized protocol, you may need separate integrations for each application.

MCP provides a common interface that compatible applications can use to communicate with MCP servers.

**Benefits:**

- Standardized communication.
- Reusable integrations.
- Easier connection to external services.
- Separation between AI applications and external systems.
- Supports different tools and data sources.

---

## 3. MCP Architecture

MCP follows a client-server architecture.

There are three main components:

### 1. MCP Host

The host is the AI application that provides the environment for interacting with MCP.

Examples:
- Claude Desktop
- An AI application built with LangGraph.
- An AI coding assistant.

The host manages the overall interaction and creates MCP client connections.

### 2. MCP Client

The client is the component inside the host that communicates with an MCP server.

Responsibilities:
- Establishes a connection with an MCP server.
- Handles protocol communication.
- Discovers available capabilities.
- Sends requests and receives responses.

### 3. MCP Server

The server exposes capabilities to AI applications through MCP.

For example, a GitHub MCP server might provide:
- Repository information.
- Issue management.
- Pull request operations.
- File access.

**Important:** An MCP server does not have to be a separate physical machine. It can run locally or remotely.


---

## 4. How Does MCP Work?

Consider an AI chatbot that needs to read a file from your computer.

The workflow:

1. The user asks the chatbot to read a file.
2. The AI application identifies that an external capability is needed.
3. The MCP client communicates with the connected MCP server.
4. The server exposes available tools and their descriptions.
5. The AI application selects the appropriate tool.
6. The client sends the tool execution request to the server.
7. The server executes the operation.
8. The result is returned to the AI application.
9. The LLM processes the result and responds to the user.

![MCP communication workflow](images/mcp.png)

**Remember:** MCP standardizes communication. It does not itself decide what the AI should do. The LLM or application controls the decision-making.

## 5. Three Main MCP Server Capabilities

MCP provides three important server-side capabilities:

### A. Tools

Tools are executable functions that an AI application can invoke.

Examples:
- Search GitHub repositories.
- Execute a database query.
- Read a file.
- Send an email.
- Search the web.

Tools are generally **model-controlled**: the AI application can make them available to the model, which decides when to request them.

### B. Resources

Resources provide data or contextual information to an AI application.

Examples:
- Database records.
- Documents.
- Configuration files.
- Application information.
- Logs.

Resources are generally **application-controlled**: the application decides which resources to retrieve and provide as context.

### C. Prompts

Prompts are reusable prompt templates exposed by an MCP server.

Examples:
- Code review prompt.
- Database analysis prompt.
- Report generation prompt.
- Debugging instructions.

Prompts are generally **user-controlled**: the application can present them for the user to select and use.

---

## 6. MCP Tools vs Normal Function Calling

You have already learned tool calling in LangChain and LangGraph.

Let's understand the difference.


### MCP Tool Calling

With MCP, tools are exposed by an MCP server.

An MCP client connects to the server, discovers its available tools, and makes them accessible to the AI application.

You do not need to implement every external integration directly inside your chatbot.

**Main difference:**

- Normal tool calling: You define and register the tools yourself.
- MCP: An MCP server exposes tools through a standardized protocol, and your application connects to them.

**Important:** MCP does not replace function calling. MCP provides the communication layer for discovering and invoking external capabilities. The AI application can still use normal LLM tool calling internally.

## 7. MCP Transport Mechanisms

MCP supports different ways for clients and servers to communicate.

### A. STDIO

STDIO stands for Standard Input/Output.

The client launches a local MCP server as a subprocess and communicates through:

- Standard Input (stdin).
- Standard Output (stdout).

**Example use cases:**
- Local file system access.
- Local development tools.
- Desktop AI applications.

### B. Streamable HTTP

Streamable HTTP allows an MCP client to communicate with a remote MCP server using HTTP.

It supports:
- Request and response communication.
- Streaming responses when needed.
- Remote server connections.

**Example use cases:**
- Cloud-hosted MCP servers.
- Remote database services.
- Shared organizational tools.

**Note:** Streamable HTTP is the modern HTTP transport in MCP. Older HTTP+SSE transport implementations are legacy.
